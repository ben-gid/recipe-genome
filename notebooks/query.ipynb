```python
from pathlib import Path
import os
from pprint import pprint
from datetime import datetime

import weaviate
from weaviate.classes.config import Configure

DATA_DIR = Path(os.getcwd()).resolve().parent / "data"
DS_PATH = DATA_DIR/ "parsed-recipes"
```

```python
client = weaviate.connect_to_local()
```

```python
from weaviate.classes.query import Filter, MetadataQuery
```

```python
result = client.collections.get("Recipes").query.hybrid(
        query="healthy and cozy recipe for the winter",
        filters=Filter.by_property("calories").less_or_equal(1_000),
        limit=1,
        return_metadata=MetadataQuery(score=True, explain_score=True)
    )
```

```python
for obj in result.objects:
    pprint(dict(**obj.properties))
```
